```python
import re
import random
import string
```

```python
def check_strength(password):
    score = 0
    remarks = ""
    
    # 1. Check Length
    if len(password) >= 8:
        score += 1
    else:
        remarks += "- Too short (min 8 chars)\n"

    # 2. Check for Digits
    if re.search(r"\d", password):
        score += 1
    else:
        remarks += "- Missing a number\n"

    # 3. Check for Uppercase
    if re.search(r"[A-Z]", password):
        score += 1
    else:
        remarks += "- Missing an uppercase letter\n"

    # 4. Check for Special Characters
    if re.search(r"[!@#$%^&*()\-+]", password):
        score += 1
    else:
        remarks += "- Missing a special character\n"

    return score, remarks

def generate_strong_password(length=12):
    # Combine all possible characters
    all_chars = string.ascii_letters + string.digits + string.punctuation
    # Randomly pick characters to meet the length
    new_password = "".join(random.choice(all_chars) for _ in range(length))
    return new_password

# --- Main Logic ---
user_input = input("Enter a password to check: ")
strength_score, feedback = check_strength(user_input)

print(f"\nStrength Score: {strength_score}/4")
if strength_score < 4:
    print(f"Suggestions:\n{feedback}")
    choice = input("Would you like me to generate a strong password for you? (y/n): ")
    if choice.lower() == 'y':
        print(f"Your new password: {generate_strong_password()}")
else:
    print("That is a very strong password!")
```
